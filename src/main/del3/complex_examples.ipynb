To create a Complex number follow example bellow

```python
from src.main.del3.complex import Complex

complex1 = Complex(1, 2)
complex2 = Complex(3, 4)
```

The different ways print functions are displayed underneath

```python
print(complex1)

print([complex1, complex2])

print(complex1.re())
print(complex1.im())

print(Complex())
print(Complex(5))
```

The supported cualculations are displayed underneath

```python
complex1_plus_2 = complex1.add(complex2)
complex1_minus_2 = complex1.sub(complex2)
complex1_mul_2 = complex1.mul(complex2)
complex1_div_2 = complex1.div(complex2)

print(complex1_plus_2)
print(complex1_minus_2)
print(complex1_mul_2)
print(complex1_div_2)
```

Logical tests underneath

```python
print(complex1 == complex1)
print(complex1 == complex2)
print(complex1 != complex2)
```
